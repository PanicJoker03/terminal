## 1. Visión del producto

Terminal aspira a ser un **sistema personal de orientación y toma de decisiones**, no solamente un gestor de tareas.

Su trabajo es ayudar a responder cinco preguntas:

1. **¿Qué está pasando realmente?** — estado de proyectos, obligaciones, recursos, tiempo, energía y riesgos.
2. **¿Qué importa ahora?** — distinguir lo urgente, lo importante, lo opcional y lo que puede descartarse.

- Más autonomía económica y profesional.

### Una aclaración importante

Terminal puede reducir incertidumbre y sufrimiento evitable; no puede garantizar una vida perfecta, eliminar todas las emociones desagradables ni predecir el futuro. Su promesa debe ser **mejor orientación y mayor capacidad de respuesta**, no control absoluto sobre la realidad.

## 5. Entropy Engine — detectar desorden sin inventar precisión

«Entropía» es aquí una metáfora operativa para representar fricción y pérdida de control, no una medición física literal de la vida.

### Señales que puede detectar

- Tareas vencidas o sin siguiente acción.
- Compromisos que compiten por el mismo tiempo.
- Dependencias bloqueadas.
- Gastos recurrentes que no se han revisado.
- Riesgos sin mitigación.
- Decisiones importantes pospuestas.
- Proyectos abiertos sin criterio de éxito o cierre.
- Información obsoleta o contradictoria.
- Exceso de trabajo comprometido respecto a la capacidad disponible.

### Salida recomendada

En vez de un «índice de entropía» opaco, mostrar:

1. **Hallazgo:** qué se detectó.
2. **Evidencia:** qué datos lo respaldan.
3. **Impacto posible:** qué podría ocurrir.
4. **Confianza:** alta, media o baja.
5. **Opciones:** resolver, reducir, delegar, posponer o aceptar.
6. **Siguiente paso:** una acción concreta y proporcionada.

No sumar riesgos heterogéneos en una puntuación aparentemente científica. Si se usa una puntuación, debe explicarse su fórmula y sus límites.



## 6. Decision Engine — recomendaciones inteligentes

El motor compara alternativas teniendo en cuenta más que urgencia o productividad.

### Criterios posibles

- Alineación con objetivos.
- Beneficio económico y sostenibilidad.
- Tiempo y esfuerzo requeridos.
- Riesgo y reversibilidad.
- Autonomía que aporta o resta.
- Impacto en descanso y bienestar.
- Coste de oportunidad.
- Calidad de la evidencia.
- Incertidumbre y consecuencias de equivocarse.

### Formato de recomendación

Cada recomendación debería incluir:

- **Recomendación:** qué hacer.
- **Por qué:** razones principales.
- **Evidencia:** datos y fuentes.
- **Supuestos:** qué se está dando por cierto.
- **Alternativas:** qué más se podría hacer.
- **Coste:** tiempo, dinero, energía y riesgo.
- **Confianza:** qué tan sólida es la evaluación.
- **Condición de revisión:** qué dato nuevo cambiaría la recomendación.

### Política de decisión

- Preferir acciones pequeñas, reversibles y de alto aprendizaje cuando hay incertidumbre.
- No recomendar más trabajo por defecto.
- Incluir explícitamente «no hacer nada por ahora» cuando sea razonable.
- Pedir confirmación antes de enviar mensajes, aceptar ofertas, renunciar, gastar dinero o borrar datos.

## 8. Corporate Autonomy Engine — ENTER · EARN · EXIT

Objetivo: usar el empleo como una fuente legítima de ingresos, experiencia y opciones, sin convertirlo en el centro de la identidad personal.

### ENTRY — entrar con información

- Investigar empresa, puesto, equipo y expectativas.
- Comparar compensación total, prestaciones, horarios y estabilidad.
- Revisar contrato, propiedad intelectual, confidencialidad, exclusividad y terminación.
- Identificar señales de alerta y preguntas para el proceso.

### VALUE — aportar profesionalmente

- Acordar entregables y expectativas.
- Mantener documentación útil.
- Automatizar tareas permitidas.
- Registrar logros para el CV.
- Mantener límites de horario y alcance.
- No copiar código, datos, secretos ni propiedad de la empresa a proyectos personales.

### EARN — convertir ingresos en autonomía

- Presupuestar gastos esenciales.
- Calcular meses de gastos cubiertos por ahorro.
- Crear un fondo de emergencia realista.
- Reducir dependencia de un único ingreso.
- Financiar desarrollo de productos propios con tiempo y presupuesto explícitos.

### EXIT — salida limpia

- Revisar contrato y legislación aplicable.
- Confirmar preaviso y obligaciones.

- Mantener comunicación profesional y evidencia de la entrega.

**Principio:** autonomía no significa engañar ni abandonar compromisos. Significa poder elegir, negociar y salir de manera legal, honesta y ordenada.



## 9. Life Skip Engine — aprender sin pagar todos los costes

### Misión

Identificar experiencias potencialmente costosas y buscar maneras de obtener información, desarrollar capacidades o alcanzar objetivos sin atravesar innecesariamente el daño asociado.

No se trata de evitar toda dificultad. Se trata de evitar sufrimiento que no aporta un beneficio proporcional.

### Flujo de trabajo

1. **Detectar:** describir la decisión o experiencia que se quiere evitar.
2. **Descomponer:** identificar qué aprendizaje u objetivo se busca.
3. **Investigar:** buscar evidencia, experiencias de terceros, documentación y asesoramiento.
4. **Simular:** comparar escenarios y consecuencias plausibles.
5. **Diseñar alternativas:** mentoría, prototipo, prueba pequeña, negociación, curso, consulta profesional.
6. **Elegir:** seleccionar una ruta proporcional al riesgo y los recursos.
7. **Revisar:** comprobar qué ocurrió y actualizar el modelo.

### Ejemplos

| Lección costosa | Ruta preventiva posible |
|---|---|
| Entrar a una empresa con cultura dañina | Investigar, preguntar al equipo, revisar condiciones y señales de alerta. |
| Agotar ahorros por mala planificación | Presupuesto, escenarios de flujo de caja y límites de gasto. |
| Quemarse por exceso de trabajo | Límites, estimación de capacidad, pausas y renegociación temprana. |
| Construir un producto sin demanda | Entrevistas, prototipo y prueba de interés antes de invertir demasiado. |
| Elegir una trayectoria profesional por inercia | Experimentos de bajo coste, conversaciones y análisis de alternativas. |
| Perder datos por fallo de hardware | Backups automatizados, copia externa y pruebas de restauración. |

### Salvaguardas

- No confundir historias anecdóticas con evidencia general.
- No tratar simulaciones como predicciones seguras.
- No saltarse asesoría médica, legal o financiera especializada cuando sea necesaria.
- No evitar toda incomodidad: algunas actividades valiosas requieren esfuerzo, incertidumbre o práctica.
- Distinguir entre incomodidad innecesaria y esfuerzo elegido que sirve a un objetivo propio.

## 14. Métricas de éxito

No medir el éxito únicamente por cantidad de tareas completadas.

- Porcentaje de gastos esenciales cubiertos por ingresos estables.

- Decisiones pendientes con siguiente paso definido.

- Proporción de recomendaciones con evidencia y supuestos visibles.

### Sostenibilidad
- Compromisos activos frente a capacidad disponible.
- Frecuencia de trabajo fuera de límites elegidos.

- Valoración subjetiva de carga mental, si el usuario decide registrarla.

### Calidad del producto
- Backups completados y restauraciones verificadas.
- Errores y tiempo de recuperación.
- Acciones sensibles que requieren confirmación.
- Recomendaciones corregidas por el usuario.
- Tiempo que Terminal consume frente al tiempo que ahorra.

### Anti-métricas

No optimizar para:
- Rachas interminables.
- Uso compulsivo de la aplicación.
- Cantidad de notificaciones.
- Horas trabajadas.
- Cumplimiento ciego de recomendaciones.

## 16. Prototipo: comparar alternativas con criterios explícitos

El siguiente ejemplo es deliberadamente sencillo. Sirve para explorar una decisión, no para decidir automáticamente por una persona.

Los pesos son preferencias hipotéticas. Cambiarlos puede cambiar el resultado. Las puntuaciones son juicios orientativos, no mediciones objetivas.


````python

# Prototipo educativo de comparación de opciones.
# Escala 1-5: 5 significa más favorable.
# Para "riesgo" y "esfuerzo", 5 significa menor riesgo / menor esfuerzo.

weights = {
    "sostenibilidad_financiera": 0.25,
    "autonomia": 0.20,
    "bienestar_sostenible": 0.25,
    "aprendizaje": 0.10,
    "reversibilidad": 0.10,
    "bajo_esfuerzo": 0.10,
}

options = {
    "Aceptar empleo A": {
        "sostenibilidad_financiera": 4,
        "autonomia": 2,
        "bienestar_sostenible": 2,
        "aprendizaje": 4,
        "reversibilidad": 3,
        "bajo_esfuerzo": 2,
    },
    "Continuar búsqueda selectiva": {
        "sostenibilidad_financiera": 2,
        "autonomia": 4,
        "bienestar_sostenible": 4,
        "aprendizaje": 3,
        "reversibilidad": 5,
        "bajo_esfuerzo": 3,
    },
    "Trabajo temporal + producto propio": {
        "sostenibilidad_financiera": 3,
        "autonomia": 4,
        "bienestar_sostenible": 3,
        "aprendizaje": 4,
        "reversibilidad": 4,
        "bajo_esfuerzo": 2,
    },
}

def score_option(criteria, weights):
    missing = set(weights) - set(criteria)
    if missing:
        raise ValueError(f"Faltan criterios: {sorted(missing)}")
    if not abs(sum(weights.values()) - 1.0) < 1e-9:
        raise ValueError("Los pesos deben sumar 1.0")
    return sum(criteria[k] * weights[k] for k in weights)

ranking = sorted(
    ((name, score_option(criteria, weights)) for name, criteria in options.items()),
    key=lambda item: item[1],
    reverse=True,
)

for position, (name, score) in enumerate(ranking, start=1):
    print(f"{position}. {name}: {score:.2f}/5")

print("\nImportante: revisa los supuestos y las puntuaciones antes de interpretar el ranking.")
````
